## Overview 
This notebook works with the EngageMetrics employee dataset to clean and transform various data types. Common data cleaning challenges like inconsistent date formats, mixed case categories, and currency-formatted numbers are addressed. 

## Learning Outcomes 
- Transform date strings into standardized datetime objects
- Normalize categorical variables for consistency
- Convert currency strings to numeric values
- Handle missing values using pandas methods
- Apply data type transformations at scale

## Dataset Information 
EngageMetrics’s <b>employee_insights.csv</b> dataset, containing employee records with various fields including:
- Dates (last_training_date, last_promotion_date)
- Categorical data (department, work_mode)
- Currency values (salary)

<b>3. utils functions</b>

## Activities
### Activity 1: Initial Data Exploration 

Let's first understand our data's current state.

<b>Step 1:</b> Load and examine the data:

```python
# Load the dataset
df = pd.read_csv(employee_insights_path)

# Examine the first few rows and data info
display(df.head())

print("\nDataset Info:")
print(df.info())
```

### Activity 2: Date Standardization

<b>Step 1:</b> Examine date formats:

```python
print("Training dates sample:")
print(df['last_training_date'].head())

print("\nPromotion dates sample:")
print(df['last_promotion_date'].head())
```

<b>Step 2: </b> Convert both date columns to datetime format: 

```python
df['last_training_date'] = pd.to_datetime(df['last_training_date'], format='mixed', errors='raise')
df['last_promotion_date'] = pd.to_datetime(df['last_promotion_date'], format='mixed', errors='raise')

print(df['last_training_date'].dtype)
print(df['last_promotion_date'].dtype)
```

### Activity 3: Salary Data Cleaning
<b>Step 1:</b> Examine salary values:

```python
print("Salary values sample:")
print(df['salary'].head())
```

<b>Step 2:</b> Clean the salary column by removing currency symbols and converting to float: 

```python
df['salary'] = df['salary'].str.replace('$', '').replace(',', '').str.strip()
df['salary'] = pd.to_numeric(df['salary'])
print(df['salary'].dtype)
print(df['salary'].head())
```

### Activity 4: Categorical Data Normalization
<b>Step 1:</b> Examine the work_mode column values:

```python
print('work_mode values')
print(df['work_mode'].value_counts())
print('\n')
print('work_experiece values')
print(df['work_experience'].value_counts())
print('\n')
print('department values')
print(df['department'].value_counts())
```

<b>Step 2:</b> Standardize values to be consistent case and format: 

```python
df['department'] = df['department'].str.strip().str.lower()

df['work_experience'] = pd.to_numeric(df['work_experience'])
df['work_experience'] = df['work_experience'].str.replace('years', '').str.strip()

work_mode_mapping = {
    'remote work': 'remote', 'hybrid': 'hybrid',
    'on-site': 'on-site', 'remote': 'remote'
}

df['work_mode'] = df['work_mode'].str.strip().str.lower()
df['work_mode'] = df['work_mode'].map(work_mode_mapping)
```

```python
print('work_mode values')
print(df['work_mode'].value_counts())
print('\n')
print('department values')
print(df['department'].value_counts())

print(df['work_experience'].dtype)
print(df['work_experience'].head())
```

### Activity 5: Handle missing values

```python
# Check initial missing values
print("\nMissing values before cleaning:")
print(df.isnull().sum())
```

```python
num_cols = df.select_dtypes(include=['number']).columns
cat_cols = df.select_dtypes(include=['object']).columns

print(f'numeric cols: {list(num_cols)}')
print(f'categorical cols: {list(cat_cols)}')

df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

```python
# Verify missing values have been handled
print("\nRemaining missing values after cleaning:")
print(df.isnull().sum())
```
