## Activities
### Activity 1: Basic Array Operations 
Create and manipulate NumPy arrays using employee data.

<b>Step 1:</b> Load dataset:

```python
# Load the dataset
df = pd.read_csv(employee_insights_path)
df.head(5)
```

<b>Step 2:</b>  Create arrays from employee data:

```python
# Extract overtime_hours into a NumPy array filling NaN values as 0
overtime = df['overtime_hours'].fillna(0).to_numpy()

# Create an array from satisfaction_score column
score = df['satisfaction_score'].to_numpy()

# Verify arrays
print("Overtime array shape:", overtime.shape)
print("Sample overtime values:", overtime[:5])

print("\nSatisfaction array shape:", score.shape)
print("Sample satisfaction values:", score[:5])
```

### Activity 2: Vectorized Calculations
Compare loop-based vs. vectorized operations for calculating overtime cost.

<b>Step 1:</b> Calculate overtime pay using a traditional Python loop:

```python
# Traditional loop approach
def calculate_overtime_loop(hours, rate=20):
    total = 0
    for hour in hours:
        if not np.isnan(hour):
            total += hour * rate
    return total
```

<b>Step 2:</b>  Implement a vectorized NumPy version:

```python
def calculate_overtime_vec(hours, rate=20):
    return np.nansum(hours * rate)
```

```python
start_time = time.time()
loop_result = calculate_overtime_loop(overtime)
loop_time = time.time() - start_time

start_time = time.time()
vec_result = calculate_overtime_vec(overtime)
vec_time = time.time() - start_time

print(f"Loop result: {loop_result:.2f}")
print(f"Loop time: {loop_time:.6f} seconds")
print(f"\nVectorized result: {vec_result:.2f}")
print(f"Vectorized time: {vec_time:.6f} seconds")
print(f"\nSpeedup factor: {loop_time / vec_time:.2f}x")
```

### Activity 3: Statistical Operations 
Apply NumPy's statistical functions to analyze employee metrics.

<b>Step 1:</b> Calculate basic statistics:

```python
print(f'avg satisfaction: {np.nanmean(score):.2f}')
print(f'median satisfaction: {np.nanmedian(score):.2f}')
print(f'std satisfaction: {np.nanstd(score):.2f}')

percentiles = np.nanpercentile(score, [25, 75])
print(f"\nPercentiles:")
print(f"25th percentile: {percentiles[0]:.2f}")
print(f"75th percentile: {percentiles[1]:.2f}")

print(f"Min: {np.nanmin(score):.2f}")
print(f"Max: {np.nanmax(score):.2f}")
print(f"Range: {np.nanmax(score) - np.nanmin(score):.2f}")
```
