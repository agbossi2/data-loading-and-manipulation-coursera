## Dataset Information 
The Two EngageMetrics datasets used before:

- <b>employee_insights.csv:</b> Contains employee performance metrics, satisfaction scores, and work-related information

## Activities

### Activity 1: Data Loading and Initial Exploration

<b>Step 1:</b> Import required libraries and load datasets:

```python
import pandas as pd

# Load the datasets
education_df = pd.read_excel('education_data.xlsx')
employee_df = pd.read_csv('employee_insights.csv')
```

<b>Step 2:</b> Examine the data:

```python
# Display first few rows of each dataset
print("Education Data:")
display(education_df.head())

print("Employee Data:")
display(employee_df.head())
```

### Activity 2: Data Cleaning and Preparation
<b>Step 1:</b> Clean the work_mode column:

```python
# Standardize string values
employee_df['work_mode'] = employee_df['work_mode'].str.strip().str.lower()
employee_df['department'] = employee_df['work_mode'].str.strip().str.lower()
employee_df['educational_background'] = employee_df['work_mode'].str.strip().str.lower()

# Convert salary to numeric
employee_df['salary'] = employee_df['salary'].str.replace('$', '').replace(',', '')
employee_df['salary'] = pd.to_numeric(employee_df['salary'])
```

### Activity 3: Merging Datasets
<b>Step 1:</b> Merge education and employee datasets:

```python
merged_df = pd.merge(
    employee_df,
    education_df,
    left_on='employee_id',
    right_on='ID',
    how='inner'
)
merged_df
```

<b>Step 2:</b> Handle missing values:

```python
print('employees')
print(employee_df.isna().sum())
print('education df')
print(education_df.isna().sum())
```

```python
numeric_cols = merged_df.select_dtypes(include=['number']).columns
categoric_cols = merged_df.select_dtypes(include=['object']).columns

merged_df[numeric_cols] = merged_df[numeric_cols].fillna(merged_df[numeric_cols].median())
merged_df[categoric_cols] = merged_df[categoric_cols].fillna(merged_df[categoric_cols].mode().iloc[0])
```

### Activity 4: Analysis and Insights
<b>Step 1:</b> Analyze educational backgrounds by department:

```python
dep_bg_count = merged_df.groupby(['department', 'educational_background']) \
                        .size() \
                        .unstack(fill_value=0)

display(dep_bg_count)
```

<b>Step 2:</b> Calculate average satisfaction scores by educational background:

```python
avg_satisf = merged_df.groupby('educational_background')['satisfaction_score'].mean().reset_index()
display(avg_satisf)
```
